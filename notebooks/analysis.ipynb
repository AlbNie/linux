# Grid Strategy Analysis
Use this notebook to load results stored in the `results/` directory and compare trials.

```python
import json
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

summary_path = Path('results/summary.csv')
if summary_path.exists():
    df = pd.read_csv(summary_path)
    display(df)

equity_files = sorted(Path('results').rglob('equity_curve.csv'))
if equity_files:
    for eq_path in equity_files[:5]:
        df_eq = pd.read_csv(eq_path, parse_dates=['timestamp'])
        plt.figure(figsize=(10, 4))
        plt.plot(df_eq['timestamp'], df_eq['portfolio_btc'], label='BTC')
        plt.title(eq_path.parent.name)
        plt.legend()
        plt.show()
```
